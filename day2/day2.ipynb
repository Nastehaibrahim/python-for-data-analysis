**Filtering** **data**

Numerical filtering

```python
# Convert Tuition_Fee to numeric, coercing non-numeric string values to NaN
data['Tuition_Fee'] = pd.to_numeric(data['Tuition_Fee'], errors='coerce')

```

```python
# Now filter the DataFrame
data[data['Tuition_Fee'] < 10]
```

categorical filtering

```python
data[data['Gender']=='Female']
```

Filtering Multiple Categories

```python

data[data['Gender'].isin(['Female','NAN'])]
```

Multiple Conditions

```python
data[ (data['Gender']=='Female') & (data['Tuition_Fee']< 20) ]
```
